### Classification avec K = 3

#### Nous disposons d'un ensemble train de 7 clients et d'un jeu de test de 4 clients. Chaque client est caractérisé par deux variables explicatives: une charge (x) et un nombre d'accidents (z). La variable cible correspond au défaut de paiement: (0) si absence de défaut ;  1 lorsqu'un dégaut est observé.

```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score
```

```python
df_train = pd.DataFrame({
    "client": ["A", "B", "C", "D", "E", "F", "G"],
    "x": [3, 4, 1, 2, 5, 6, 6],
    "z": [4, 4, 3, 5, 5, 3, 4],
    "defaut": [1, 0, 0, 0, 1, 1, 1]
})

df_test = pd.DataFrame({
    "client": ["P", "Q", "R", "S"],
    "x": [3, 5, 1, 6],
    "z": [3, 4, 4, 4.5],
    "defaut": [0, 1, 1, 0]
})
```

```python
display(df_train)
```

```python
display(df_test)
```

```python
X_train = df_train[["x", "z"]]
y_train = df_train["defaut"]

X_test = df_test[["x", "z"]]
y_test = df_test["defaut"]
```

```python
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)
```

```python
y_pred = knn.predict(X_test)
```

```python
df_test["y_pred"] = y_pred
df_test
```

```python
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
```

```python
print("Accuracy: ", accuracy)
print("Précision: ", precision)
print("Recall: ", recall)
```

Accuracy: Sur l'ensemble des clients, notre modèle a correctement prédis la situation de 50%.

Précision: La précision signifie que, parmi les clients prédits en défaut par le modèle, 50% sont réellement en défaut.

Recall: Le recall de 50% signifie que le modèle à correctement identifié 50% des clients réellement en défaut.
