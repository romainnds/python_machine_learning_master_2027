###Classification avec une variable, K= 1

#### Une banque cherche à repérer des opérations suspectes (classe 1) à partir d'un score fictif. La classe 0 désigne une opération normale.

```python
X_train = [1, 3, 5, 7, 9]
y_train = [0, 0, 1, 1, 0]

X_test = [2.2, 4.6, 6.6, 8.6]
y_test = [0, 1, 1, 1]
```

Question 1: Pour chaque opération de test, identifiez le voisin le plus proche avec d(x, x') = |x - x'| et prédiser sa classe.

```python
y_pred = []
```

```python
for i in range(len(X_test)):
    print("X_test: ", X_test[i])
    distances = []

    for j in range(len(X_train)):
        distance = abs(X_test[i] - X_train[j])
        distances.append(distance)
        print("Distance: ", round(distance, 2))

    distance_min = min(distances)
    indice = distances.index(distance_min)

    prediction = y_train[indice]
    y_pred.append(prediction)

    print("Distance minimale: ", round(distance_min, 2))
    print("Classe prédite: ", prediction)
    print()

print("Prédictions finales: ", y_pred)
```

Question 2: Déterminez le nombre de bonnes prédictions et le nombre de cas suspects effectivement détectés.

```python
bonnes_pred = 0
suspects_detectes = 0

for i in range(len(y_test)):

  if y_pred[i] == y_test[i]:
    bonnes_pred +=1
  if y_test[i] == 1 and y_pred[i] == 1:
    suspects_detectes += 1

print("Bonnes prédictions: ", bonnes_pred)
print("Cas suspects détectés: ", suspects_detectes)
```

Question 3: Calculez l'accuracy et le recall.

```python
accuracy = bonnes_pred / len(y_test)
print("Accuracy: ", round(accuracy * 100, 2), "%")
```

Le modèle a correctement classé 75% des opérations de test.

```python
total_suspects = y_test.count(1)
recall = suspects_detectes / total_suspects
print("Rappel: ", round(recall * 100, 2), "%")
```

Le modèle détecte enviro 66.67% des opérations réellement suspectes.

#### Une version plus simple et rapide avec le module Sklearn.

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, recall_score

knn = KNeighborsClassifier(n_neighbors=1)

#Train
knn.fit([[x] for x in X_train], y_train)

#Pred
predictions_sklearn = knn.predict([[x] for x in X_test])

#Accuracy et recall
accuracy_sklearn = accuracy_score(y_test, predictions_sklearn)
rappel_sklearn = recall_score(y_test, predictions_sklearn)

print("Prédictions: ", predictions_sklearn)
print("Accuracy: ", round(accuracy_sklearn * 100, 2), "%")
print("Recall: ", round(rappel_sklearn * 100, 2), "%")
```
