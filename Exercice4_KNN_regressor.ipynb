```python
import pandas as pd
import numpy as np
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error
from sklearn.metrics.pairwise import euclidean_distances
```

```python
train = pd.DataFrame({
    "sinistre": ["A","B","C","D","E","F"],
    "x": [1, 2, 1, 4, 5, 4],
    "z": [1, 1, 2, 4, 4, 5],
    "montant":[900, 1200, 1500, 3000, 3600, 4200]
})

test = pd.DataFrame({
    "sinistre": ["P","Q","R"],
    "x": [1, 4.5, 2],
    "z": [1.5, 4, 2],
    "montant": [1500, 4200, 2100]
})
```

```python
train
```

```python
test
```

```python
X_train = train[["x", "z"]]
y_train = train["montant"]

X_test = test[["x", "z"]]
y_test = test["montant"]
```

```python
knn = KNeighborsRegressor(n_neighbors=3)
knn.fit(X_train, y_train)
```

```python
df_distances = pd.DataFrame(euclidean_distances(X_test, X_train, squared=True), index=test["sinistre"], columns=train["sinistre"])
df_distances
```

Les plus proches voisins de:

*   P: A, B et C // Avec un Ypred de (900+1200+1500)/3 = 1200
*   Q: D, E et F // Avec un Ypred de (3000+3600+4200)/3 = 3600
*   R: A, B et C // Avec un Ypred de 1200.

```python
y_pred = knn.predict(X_test)
```

```python
y_pred
```

```python
test["prediction"] = y_pred
```

```python
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
```

```python
print("MAE: ", mae)
print("RMSE: ", round(rmse, 2))
```

MAE: Nos prédictions s'écartent en moyenne de 600 du montant réel.
RMSE: Contrairement à la MAE, on met les erreurs au carré avant de calculer la moyenne. Le RMSE pénalise davantage les grandes erreurs que la MAE.
